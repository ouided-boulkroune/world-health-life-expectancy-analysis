```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Life Expectancy Data.csv')
print("Dataset loaded successfully!")
```

```python
df.head(10)
```

```python
print("Rows and columns:", df.shape)
print("\nColumn names:\n", df.columns.tolist())
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.columns = df.columns.str.strip()
print("Column names cleaned!")
print(df.columns.tolist())
```

```python
numeric_cols = df.select_dtypes(include='number').columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
print("Missing values handled!")
df.isnull().sum()
```

```python
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

```python
print("Clean dataset shape:", df.shape)
print("Missing values remaining:", df.isnull().sum().sum())
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Status', y='Life expectancy', data=df, palette='Set2')
plt.title('Life Expectancy: Developed vs Developing Countries')
plt.xlabel('Country Status')
plt.ylabel('Life Expectancy (years)')
plt.tight_layout()
plt.savefig('chart1_status_vs_life_expectancy.png')
plt.show()
```

```python
top10 = df.groupby('Country')['Life expectancy'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top10.plot(kind='bar', color='steelblue')
plt.title('Top 10 Countries by Average Life Expectancy')
plt.xlabel('Country')
plt.ylabel('Average Life Expectancy (years)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart2_top10_countries.png')
plt.show()
```

```python
bottom10 = df.groupby('Country')['Life expectancy'].mean().sort_values(ascending=True).head(10)

plt.figure(figsize=(10,6))
bottom10.plot(kind='bar', color='tomato')
plt.title('Bottom 10 Countries by Average Life Expectancy')
plt.xlabel('Country')
plt.ylabel('Average Life Expectancy (years)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart3_bottom10_countries.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='GDP', y='Life expectancy', data=df, hue='Status', alpha=0.6)
plt.title('GDP vs Life Expectancy')
plt.xlabel('GDP')
plt.ylabel('Life Expectancy (years)')
plt.tight_layout()
plt.savefig('chart4_gdp_vs_life_expectancy.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Schooling', y='Life expectancy', data=df, hue='Status', alpha=0.6)
plt.title('Schooling vs Life Expectancy')
plt.xlabel('Years of Schooling')
plt.ylabel('Life Expectancy (years)')
plt.tight_layout()
plt.savefig('chart5_schooling_vs_life_expectancy.png')
plt.show()
```

```python
import sqlite3
print("SQLite ready!")
```

```python
conn = sqlite3.connect('health_data.db')
df.to_sql('health', conn, if_exists='replace', index=False)
print("Database created successfully!")
```

```python
query1 = """
SELECT Country, ROUND(AVG("Life expectancy"), 1) AS avg_life_expectancy
FROM health
GROUP BY Country
ORDER BY avg_life_expectancy DESC
LIMIT 10
"""
result1 = pd.read_sql(query1, conn)
print("Top 10 Countries - Highest Life Expectancy:")
print(result1)
```

```python
query2 = """
SELECT Country, ROUND(AVG("Life expectancy"), 1) AS avg_life_expectancy
FROM health
GROUP BY Country
ORDER BY avg_life_expectancy ASC
LIMIT 10
"""
result2 = pd.read_sql(query2, conn)
print("Bottom 10 Countries - Lowest Life Expectancy:")
print(result2)
```

```python
query3 = """
SELECT Year, ROUND(AVG("Life expectancy"), 1) AS avg_life_expectancy
FROM health
GROUP BY Year
ORDER BY Year ASC
"""
result3 = pd.read_sql(query3, conn)
print("Global Life Expectancy by Year:")
print(result3)
```

```python
query4 = """
SELECT Status, 
       ROUND(AVG("Life expectancy"), 1) AS avg_life_expectancy,
       ROUND(AVG("Schooling"), 1) AS avg_schooling,
       ROUND(AVG("GDP"), 1) AS avg_gdp
FROM health
GROUP BY Status
"""
result4 = pd.read_sql(query4, conn)
print("Developed vs Developing - Full Comparison:")
print(result4)
```

```python
plt.figure(figsize=(10,5))
plt.plot(result3['Year'], result3['avg_life_expectancy'], 
         marker='o', color='steelblue', linewidth=2.5, markersize=6)
plt.fill_between(result3['Year'], result3['avg_life_expectancy'], 
                  alpha=0.1, color='steelblue')
plt.title('Global Life Expectancy Trend (2000-2015)', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Average Life Expectancy (years)')
plt.xticks(result3['Year'], rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('chart6_trend_2000_2015.png')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14,5))

metrics = ['avg_life_expectancy', 'avg_schooling', 'avg_gdp']
titles = ['Life Expectancy (years)', 'Schooling (years)', 'GDP']
colors = ['steelblue', 'tomato']

for i, (metric, title) in enumerate(zip(metrics, titles)):
    axes[i].bar(result4['Status'], result4[metric], color=colors)
    axes[i].set_title(title, fontsize=11)
    axes[i].set_xlabel('Country Status')
    for j, val in enumerate(result4[metric]):
        axes[i].text(j, val + (val*0.02), str(val), 
                    ha='center', fontweight='bold', fontsize=10)

plt.suptitle('Developed vs Developing Countries Comparison', 
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('chart7_developed_vs_developing.png')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16,6))

axes[0].barh(result1['Country'], result1['avg_life_expectancy'], color='steelblue')
axes[0].set_title('Top 10 Countries - Highest Life Expectancy', fontsize=11)
axes[0].set_xlabel('Average Life Expectancy (years)')
for i, val in enumerate(result1['avg_life_expectancy']):
    axes[0].text(val - 2, i, str(val), va='center', 
                color='white', fontweight='bold')

axes[1].barh(result2['Country'], result2['avg_life_expectancy'], color='tomato')
axes[1].set_title('Bottom 10 Countries - Lowest Life Expectancy', fontsize=11)
axes[1].set_xlabel('Average Life Expectancy (years)')
for i, val in enumerate(result2['avg_life_expectancy']):
    axes[1].text(val - 2, i, str(val), va='center', 
                color='white', fontweight='bold')

plt.tight_layout()
plt.savefig('chart8_top_bottom_countries.png')
plt.show()
```
